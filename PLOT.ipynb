```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

MAGNETIZACION

```python
f1='C:/Users/20pab/GF/3GF/FC/VoluntarioKawasaki/16NULA/Magnetizacion2.dat'
f2='C:/Users/20pab/GF/3GF/FC/VoluntarioKawasaki/24NULA/Magnetizacion2.dat'
f3='C:/Users/20pab/GF/3GF/FC/VoluntarioKawasaki/32NULA/Magnetizacion2.dat'

d1=pd.read_csv(f1, delimiter=',', header=0, names=['up','down','T','L','N'])
d2=pd.read_csv(f2, delimiter=',', header=0, names=['up','down','T','L','N'])
d3=pd.read_csv(f3, delimiter=',', header=0, names=['up','down','T','L','N'])

plt.scatter(d1['T'], d1['up'], linewidth=2, marker='o', s=10, c='r', label='L16up')
plt.scatter(d1['T'], -d1['down'], linewidth=2, marker='o', s=10, c='coral', label='L16down')
plt.plot(d2['T'], d2['up']-d2['down'], linewidth=2, c='purple', label='Magnetizacion total')
plt.scatter(d2['T'], d2['up'], linewidth=2, marker='o', s=10, c='b', label='L24up')
plt.scatter(d2['T'], -d2['down'], linewidth=2, marker='o', s=10, c='lightblue', label='L24down')
plt.scatter(d3['T'], d3['up'], linewidth=2, marker='o', s=10, c='green', label='L32up')
plt.scatter(d3['T'], -d3['down'], linewidth=2, marker='o', s=10, c='lightgreen', label='L32down')

plt.xlabel(r'$T$')
plt.ylabel(r'$M$')
# Crear una leyenda personalizada para ambas curvas con la misma etiqueta
handles, labels = plt.gca().get_legend_handles_labels()
unique_labels = [r'$L=16$', r'$L=24$', r'$L=32$', 'Magnetizacion total']
unique_handles = [handles[0], handles[3], handles[5], handles[2]]  # Puedes usar handles[0] o handles[1], ya que ambos representan una curva

# Mostrar la leyenda personalizada
plt.legend(unique_handles, unique_labels)
#plt.grid()
plt.show()
```

ENERGIA MEDIA

```python
import seaborn as sns

# Cargar el archivo con el delimitador correcto
file_path = 'C:/Users/20pab/GF/3GF/FC/VoluntarioKawasaki/16NULA/Energia.txt'
data = np.loadtxt(file_path, delimiter=',')

# Crear el mapa de calor
plt.figure(figsize=(10, 8))
sns.heatmap(data, cmap='viridis')
plt.title('Mapa de Calor de la Matriz de Energía')
plt.xlabel('Columnas')
plt.ylabel('Filas')
plt.show()
```
